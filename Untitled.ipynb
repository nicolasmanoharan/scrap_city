```python

service = FirefoxService(GeckoDriverManager().install())
driver = webdriver.Firefox(service=service)

try:
    # Ouvrir Google
    driver.get("https://www.google.com")

    # Si besoin, accepter la politique de cookies
    try:
        consent_button = WebDriverWait(driver, 5).until(
            EC.element_to_be_clickable((By.XPATH, "/html/body/div[2]/div[2]/div[3]/span/div/div/div/div[3]/div[1]/button[2]/div"))
        )
        consent_button.click()
    except Exception:
        pass

    # Attendre que la barre de recherche soit présente
    search_box = WebDriverWait(driver, 10).until(
        EC.presence_of_element_located((By.NAME, "q"))
    )

    # Effectuer la recherche "entreprise YOURART"
    query = "entreprise YOURART"
    search_box.clear()
    search_box.send_keys(query)
    search_box.send_keys(Keys.RETURN)

    # Attendre que les résultats s'affichent
    WebDriverWait(driver, 10).until(
        EC.presence_of_all_elements_located((By.CSS_SELECTOR, "div.g"))
    )

    # Chercher le premier lien contenant "linkedin.com"
    linkedin_link = None
    links = driver.find_elements(By.XPATH, "//a[contains(@href, 'linkedin.com')]")
    if links:
        linkedin_link = links[0]
        print("Lien LinkedIn trouvé :", linkedin_link.get_attribute("href"))
        linkedin_link.click()
    else:
        print("Aucun lien LinkedIn trouvé pour cette recherche.")

    # Laisser le temps d'afficher la page LinkedIn
    time.sleep(5)

finally:
    driver.quit()
```
